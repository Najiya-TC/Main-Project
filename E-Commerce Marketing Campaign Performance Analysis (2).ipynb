<h1 style="color:#4338CA;">OUTLIERS DETECTION USING IQR METHOD</h1>



```python
# Select numerical columns and exclude the ID column
numeric_cols = df.select_dtypes(include=np.number).columns.drop('id')

numeric_cols
```

```python
# Detect potential outliers using the IQR method

outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    outlier_summary.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

### Outlier Count and Percentage

The number and percentage of potential outliers are calculated for each numerical variable using the IQR method. This helps to understand the extent of extreme values in the dataset before deciding whether any outlier treatment is required.

```python
# Calculate outlier count and percentage for each numerical variable

outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_count = len(outliers)
    outlier_percentage = (outlier_count / len(df)) * 100
    
    outlier_summary.append({
        'Column': col,
        'Outlier Count': outlier_count,
        'Outlier Percentage': round(outlier_percentage, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

### Outlier Handling Using IQR Capping

The IQR method identified potential outliers in several numerical variables. Since extreme values in e-commerce marketing data may represent genuine observations, such as high advertising spend, high-value customers, or successful campaigns, these records were not removed.

Instead, IQR-based capping was applied to reduce the influence of extreme values while retaining all observations in the dataset.

```python
# Handle outliers using IQR capping

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)
```

```python
# Verify remaining potential outliers

outlier_check = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    outlier_check.append({
        'Column': col,
        'Remaining Outliers': len(outliers)
    })

outlier_check = pd.DataFrame(outlier_check)

outlier_check
```

### Outlier Analysis Summary

The IQR method was used to identify and handle potential outliers in the numerical variables. IQR-based capping was applied to the numerical variables, excluding the id column, to reduce the influence of extreme values while retaining all observations.

After capping, the verification results show 0 remaining potential outliers across the numerical variables. This indicates that the identified extreme values were successfully brought within the IQR boundaries without removing any records from the dataset.
